```python
import pickle
import numpy as np

sfreq = 128
window_size = 2 * sfreq
step = sfreq

X = []
y = []
groups = []

for i in range(1, 33):
    file_path = f"deap_data/deap-dataset/data_preprocessed_python/s{i:02d}.dat"

    with open(file_path, 'rb') as f:
        data = pickle.load(f, encoding='latin1')

    eeg = data['data'][:, :32, :]
    labels = data['labels']

    eeg = eeg[:, :, 3*sfreq:]

    for trial_idx, trial in enumerate(eeg):
        for start in range(0, trial.shape[1] - window_size, step):
            seg = trial[:, start:start+window_size]

            X.append(seg)
            y.append(1 if labels[trial_idx, 0] > 5 else 0)
            groups.append(i)

X = np.array(X, dtype=np.float32)
y = np.array(y)

X = (X - X.mean()) / (X.std() + 1e-8)

print(X.shape, y.shape)
```

```python
import numpy as np

data = np.load("processed_eeg.npz")

X = data["X"]
y = data["y"]
```

```python
import torch

X_cnn = X.reshape(-1, 1, 32, 256)

X_cnn = torch.tensor(X_cnn, dtype=torch.float32)
y_cnn = torch.tensor(y, dtype=torch.float32)
```

```python
import torch.nn as nn

class EEGNet(nn.Module):
    def __init__(self):
        super().__init__()

        self.block1 = nn.Sequential(
            nn.Conv2d(1, 16, (1, 64), padding=(0,32)),
            nn.BatchNorm2d(16),
            nn.ELU(),
            nn.Dropout(0.25)
        )

        self.block2 = nn.Sequential(
            nn.Conv2d(16, 32, (32,1)),
            nn.BatchNorm2d(32),
            nn.ELU(),
            nn.Dropout(0.25)
        )

        self.pool = nn.AdaptiveAvgPool2d((1,1))
        self.fc = nn.Linear(32, 1)

    def forward(self, x):
        x = self.block1(x)
        x = self.block2(x)
        x = self.pool(x)
        x = x.view(x.size(0), -1)
        return self.fc(x)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = EEGNet().to(device)

pos_weight = (y == 0).sum() / (y == 1).sum()
pos_weight = torch.tensor(pos_weight, dtype=torch.float32).to(device)

criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)
optimizer = torch.optim.Adam(model.parameters(), lr=0.0005)

batch_size = 64
```

```python
for epoch in range(15):
    model.train()
    total_loss = 0

    for i in range(0, len(X_cnn), batch_size):
        xb = X_cnn[i:i+batch_size].to(device)
        yb = y_cnn[i:i+batch_size].to(device)

        optimizer.zero_grad()

        out = model(xb).squeeze()
        loss = criterion(out, yb.float())

        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / (len(X_cnn)//batch_size)

    print(f"Epoch {epoch+1}, Loss: {avg_loss:.4f}")
```

```python
from sklearn.metrics import classification_report, f1_score

model.eval()

all_logits = []
with torch.no_grad():
    for i in range(0, len(X_cnn), batch_size):
        xb = X_cnn[i:i+batch_size].to(device)
        out = model(xb).squeeze()
        all_logits.append(out.cpu())

logits = torch.cat(all_logits)
probs = torch.sigmoid(logits).numpy()

best_t = 0.5
preds = (probs > best_t).astype(int)

print("F1:", f1_score(y, preds))
print(classification_report(y, preds))
```
